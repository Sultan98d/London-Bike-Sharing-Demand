```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
df["is_holiday"] = df["is_holiday"].astype("bool")
df["is_weekend"] = df["is_weekend"].astype("bool")
```

```python
import matplotlib.pyplot as plt

plt.hist(df["cnt"], bins=30)
plt.xlabel("Bike Rentals")
plt.ylabel("Frequency")
plt.title("Distribution of Bike Sharing Demand")
plt.show()
```

```python
hourly_demand = df.groupby("hour")["cnt"].mean()

hourly_demand.plot(kind="line", marker="o", figsize=(10, 5))

plt.title("Average Bike Demand by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average Bike Rentals")
plt.xticks(range(24))
plt.grid(alpha=0.3)
plt.show()
```

```python
weekend_demand = df.groupby(["hour", "is_weekend"])["cnt"].mean().unstack()

weekend_demand.columns = ["Weekday", "Weekend"]

weekend_demand.plot(kind="line", figsize=(10, 5))

plt.title("Hourly Demand: Weekday vs Weekend")
plt.xlabel("Hour of Day")
plt.ylabel("Average Bike Rentals")
plt.xticks(range(24))
plt.show()
```

```python
seasonal_demand = df.groupby("season", observed=True)["cnt"].mean()

seasonal_demand.index = ["Spring", "Summer", "Fall", "Winter"]

seasonal_demand.plot(kind="bar", figsize=(8, 5))

plt.title("Average Bike Demand by Season")
plt.xlabel("Season")
plt.ylabel("Average Bike Rentals")
plt.xticks(rotation=0)
plt.show()
```

```python
weather_demand = df.groupby("weather_code", observed=True)["cnt"].mean()

weather_demand = weather_demand.rename(index={
    1: "Clear",
    2: "Few Clouds",
    3: "Broken Clouds",
    4: "Cloudy",
    7: "Rain",
    10: "Thunderstorm",
    26: "Snow"
})

weather_demand.sort_values().plot(kind="barh", figsize=(9, 5))

plt.title("Average Bike Demand by Weather")
plt.xlabel("Average Bike Rentals")
plt.ylabel("Weather Condition")
plt.show()
```

```python
df[["cnt", "t1", "t2", "hum", "wind_speed"]].corr()["cnt"]
```

```python
monthly_demand = df.set_index("timestamp")["cnt"].resample("MS").mean()

monthly_demand.plot(kind="line", marker="o", figsize=(12, 5))

plt.title("Monthly Bike Demand Trend")
plt.xlabel("Date")
plt.ylabel("Average Hourly Bike Rentals")
plt.show()
```

```python
total_rentals = df["cnt"].sum()
average_rentals = df["cnt"].mean()
peak_rentals = df["cnt"].max()

print("Total Rentals:", total_rentals)
print("Average Hourly Rentals:", round(average_rentals))
print("Peak Hourly Rentals:", peak_rentals)
```
